```python
%matplotlib inline

# scientific computing and plotting
import numpy as np
import pandas as pd
import xarray as xr
import matplotlib.pyplot as plt
import seaborn as sns

# HDDM related packages
import pymc as pm
import hddm
import kabuki
import arviz as az
print("The current HDDM version is: ", hddm.__version__)
print("The current kabuki version is: ", kabuki.__version__)
print("The current PyMC version is: ", pm.__version__)
print("The current ArviZ version is: ", az.__version__)

import warnings
warnings.filterwarnings('ignore')
```

```python
# read data
data = pd.read_csv('MS_categ_id_hddm_stim.csv')
data
```

```python
# accuracy coding
data['choice'] = data['response'].copy()                       # 保存 choice 编码（1=Self, 0=Other）
data['acc'] = (data['response'] == data['stim']).astype(int)   # 计算正确率

data['response'] = data['acc']   
```

```python
#hddm 
#hddm_acc
# HDDM accuracy coding 
# v varies across different stim and id

data['response'] =data['acc']
hddm_id_m0_acc_diff = hddm.HDDM(data, include=['a', 'v', 't','z'], depends_on={'v': ['id','stim']})

hddm_id_m0_acc_diff_idata = hddm_id_m0_acc_diff.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddm_id_m0_acc_diff.db', db = 'pickle')
hddm_id_m0_acc_diff.save('hddm_id_m0_acc_diff.hddm')
# hddm.load('hddm_id_m0_acc.hddm')
```

```python
az.summary(hddm_id_m0_acc_diff_idata)
```

```python
hddm_id_m0_acc_stats = hddm_id_m0_acc_diff.gen_stats()
hddm_id_m0_acc_stats[hddm_id_m0_acc_stats.index.isin(['a', 'v(Self.1)', 'v(Other.1)', 
                                                'v(Self.0)', 'v(Other.0)','z'])]
```

```python
#hddm_choice
# HDDM choice coding 
# v varies across different stim and id
hddm_id_m1_choice_diff = hddm.HDDM(data, include=['a', 'v', 't','z'], depends_on={'v': ['id', 'stim']})

hddm_id_m1_choice_diff_idata = hddm_id_m1_choice_diff.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddm_id_m1_choice_diff.db', db = 'pickle')
hddm_id_m1_choice_diff.save('hddm_id_m1_choice_diff.hddm')
# hddm.load('hddm_id_m1_choice.hddm')
```

```python
az.summary(hddm_id_m1_choice_diff_idata)
```

```python
hddm_id_m1_choice_stats = hddm_id_m1_choice_diff.gen_stats()
hddm_id_m1_choice_stats[hddm_id_m1_choice_stats.index.isin(['a', 'v(Self.1)', 'v(Other.1)', 
                                                'v(Self.0)', 'v(Other.0)','z'])]
```

```python
#HDDMStimCoding
#choice coding + v flip + dc
#accuracy coding + z flip
#hddmStim_v_flip
data['response'] = data['choice']

# stimCoding v flip + dc
hddmStim_vflip_dc_m1 = hddm.HDDMStimCoding(
    data,
    stim_col='stim',
    split_param='v',
    drift_criterion=True,
    depends_on={'v':['stim'],'dc': ['id']},
    bias =True,
    include=['v', 'a', 't', 'z'],
)

hddmStim_vflip_dc_m1_idata = hddmStim_vflip_dc_m1.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddmStim_vflip_dc_m1.db', db = 'pickle')
hddmStim_vflip_dc_m1.save('hddmStim_vflip_dc_m1.hddm')
```

```python
# model summary
az.summary(hddmStim_vflip_dc_m1_idata)
```

```python
data['response'] = data['acc']

# stimCoding z flip 
hddmStim_zflip_m1_diff = hddm.HDDMStimCoding(
    data,
    stim_col='stim',
    split_param='z',
    #drift_criterion=True,
    depends_on={'v':['id','stim']},
    bias =True,
    include=['v', 'a', 't'],
)

hddmStim_zflip_m1_diff_idata = hddmStim_zflip_m1_diff.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddmStim_zflip_m1_diff.db', db = 'pickle')
hddmStim_zflip_m1_diff.save('hddmStim_zflip_m1_diff.hddm')
```

```python
#HDDMRegression
#accuracy coding
#choice coding
#stim coding: accuracy coding + v flip + dc
#stim coding: choice coding + z flp
# accuracy coding
data['response'] = data['acc']

v_reg = [{"model": "v ~ 1", "link_func": lambda x: x}]

hddmReg_id_m0_acc_diff = hddm.HDDMRegressor(
    data,
    v_reg,
    bias=True, p_outlier=0.05,
    include=['v', 'a', 't', 'z'],
    depends_on={'v':['id','stim']},
    group_only_regressors=False
)

hddmReg_id_m0_acc_diff_idata = hddmReg_id_m0_acc_diff.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddmReg_id_m0_acc_diff.db', db = 'pickle')
hddmReg_id_m0_acc_diff.save('hddmReg_id_m0_acc_diff.hddm')
```

```python
#hddmReg_stim_v_flip
#type1：reg
# using stim coding with HDDMRegressor
# type1: choice coding + v flip +dc

data['response'] = data['choice']

# recode stimulus and apply parameter flipping
data['stim2'] = data['stim'].map({1: 1, 0: -1})

v_reg = [{'model': "v ~ 1 + id*stim2", 'link_func': lambda x: x}] 


hddmReg_stim_vflip_dc_m2 = hddm.HDDMRegressor(
    data,
    v_reg,
    bias=True,
    include=['v', 'a', 't', 'z'],
    depends_on={'v': ['id']},
    group_only_regressors=False
)

hddmReg_stim_vflip_dc_m2_idata = hddmReg_stim_vflip_dc_m2.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddmReg_stim_vflip_dc_m2.db', db = 'pickle')
hddmReg_stim_vflip_dc_m2.save('hddmReg_stim_vflip_dc_m2.hddm')
```

```python
summary = az.summary(hddmReg_stim_vflip_dc_m2_idata)
group_params = summary[~summary.index.str.contains('_subj.')]
print(group_params)
```

```python
#type2：link function
def v_link_func(x, data=data):
    # 1 is big; 0 is small
    stim = data.stim.loc[x.index]
    v_flip = x.where(stim == 1, -x)
    return v_flip

v_reg = [{'model': "v ~ 0 + id*C(stim)", 'link_func': v_link_func}] 

data['response'] = data['choice']

hddmReg_stim_vflip_m3_diff = hddm.HDDMRegressor(
    data,
    v_reg,
    bias=True, p_outlier=0.05,
    include=['v', 'a', 't', 'z'],
    group_only_regressors=False
)

hddmReg_stim_vflip_m3_diff_idata = hddmReg_stim_vflip_m3_diff.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddmReg_stim_vflip_m3_diff.db', db = 'pickle')
hddmReg_stim_vflip_m3_diff.save('hddmReg_stim_vflip_m3_diff.hddm')
```

```python
summary = az.summary(hddmReg_stim_vflip_m3_diff_idata)
group_params = summary[~summary.index.str.contains('_subj.')]
print(group_params)
```

```python
#hddmReg_stim_z_flip
# using stim coding with HDDMRegressor
# type2: acc coding + z flip 
data['response'] = data['acc']

def z_link_func(x, data=data):
  # 1 is self; 0 is other
  stim = data.stim.loc[x.index]
    z_flip = (stim * x) + (1 - x) * (1 - stim)
    return z_flip

z_reg = [{'model': "z ~ 1", 'link_func': z_link_func}] 

hddmReg_stim_zflip_m2_diff = hddm.HDDMRegressor(
    data,
    z_reg,
    bias=True, p_outlier=0.05,
    include=['v', 'a', 't', 'z'],
    depends_on={'v': ['id','stim']},
    group_only_regressors=False
)

hddmReg_stim_zflip_m2_diff_idata = hddmReg_stim_zflip_m2_diff.sample(2000, burn=500, chains=4, return_infdata=True, dbname = 'hddmReg_stim_zflip_m2_diff.db', db = 'pickle')
hddmReg_stim_zflip_m2_diff.save('hddmReg_stim_zflip_m2_diff.hddm')
```
